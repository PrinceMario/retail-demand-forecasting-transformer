```python
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

plot_acf(train_diff, lags=30, ax=axes[0])
axes[0].set_title("ACF — First-Differenced Training Series")

plot_pacf(train_diff, lags=30, ax=axes[1], method="ywm")
axes[1].set_title("PACF — First-Differenced Training Series")

plt.tight_layout()
plt.show()
```

```python
max_lag = 30

acf_values, acf_confint = acf(
    train_diff, nlags=max_lag, alpha=0.05
)
pacf_values, pacf_confint = pacf(
    train_diff, nlags=max_lag, alpha=0.05, method="ywm"
)

significant_acf_lags = []
significant_pacf_lags = []

for lag in range(1, max_lag + 1):
    acf_low, acf_high = acf_confint[lag] - acf_values[lag]
    if not (acf_low <= 0 <= acf_high):
        significant_acf_lags.append(lag)

    pacf_low, pacf_high = pacf_confint[lag] - pacf_values[lag]
    if not (pacf_low <= 0 <= pacf_high):
        significant_pacf_lags.append(lag)

print("Significant ACF lags :", significant_acf_lags)
print("Significant PACF lags:", significant_pacf_lags)
```

## 5. Supervised Sequence Construction

```python
def create_aligned_lag_samples(series, lags, start_index=24):
    series = np.asarray(series, dtype=np.float32)
    X, y = [], []

    for i in range(start_index, len(series)):
        X.append([series[i - lag] for lag in lags])
        y.append(series[i])

    return np.asarray(X, dtype=np.float32), np.asarray(y, dtype=np.float32)


def create_test_lag_samples(train_series, test_series, lags):
    train_series = np.asarray(train_series, dtype=np.float32)
    test_series = np.asarray(test_series, dtype=np.float32)

    history = np.concatenate([train_series, test_series])
    train_len = len(train_series)

    X_test = []
    for j in range(len(test_series)):
        t = train_len + j
        X_test.append([history[t - lag] for lag in lags])

    return np.asarray(X_test, dtype=np.float32)


X_train_combined, y_train_combined = create_aligned_lag_samples(
    train_scaled, combined_lags, start_index=24
)
X_test_combined = create_test_lag_samples(
    train_scaled, test_scaled, combined_lags
)

print("Primary sparse training shape:", X_train_combined.shape)
print("Primary sparse test shape    :", X_test_combined.shape)
```

## 6. Vaticinator Architecture

### Replicated architecture

**Selected lag values → Linear embedding (1→32) → Lag-aware sinusoidal PE → Transformer Encoder ×3 → final token → Linear (32→1) → next-month forecast**

**Important:** sparsity is created by **input construction**. There is no separate sparse-attention mask.

```python
class LagAwarePositionalEncoding(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.d_model = d_model

    def forward(self, x, lags):
        device = x.device

        lag_positions = torch.tensor(
            lags, dtype=torch.float32, device=device
        ).unsqueeze(1)

        div_term = torch.exp(
            torch.arange(0, self.d_model, 2, device=device).float()
            * (-math.log(10000.0) / self.d_model)
        )

        pe = torch.zeros(len(lags), self.d_model, device=device)
        pe[:, 0::2] = torch.sin(lag_positions * div_term)
        pe[:, 1::2] = torch.cos(lag_positions * div_term)

        return x + pe.unsqueeze(0)


class Vaticinator(nn.Module):
    def __init__(
        self,
        lags,
        d_model=32,
        nhead=8,
        num_layers=3,
        dim_feedforward=128
    ):
        super().__init__()

        self.lags = list(lags)
        self.embedding = nn.Linear(1, d_model)
        self.positional_encoding = LagAwarePositionalEncoding(d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            activation="relu",
            batch_first=True
        )

        self.encoder = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.output = nn.Linear(d_model, 1)

    def forward(self, x):
        x = self.embedding(x)
        x = self.positional_encoding(x, self.lags)
        x = self.encoder(x)
        final_representation = x[:, -1, :]
        return self.output(final_representation).squeeze(-1)


model_preview = Vaticinator(
    lags=combined_lags,
    d_model=CONFIG["d_model"],
    nhead=CONFIG["nhead"],
    num_layers=CONFIG["num_layers"],
    dim_feedforward=CONFIG["dim_feedforward"]
)

print(model_preview)
```

## 7. Primary 9-Lag Vaticinator — Training & Held-Out Evaluation

```python
def regression_metrics(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    mpe = np.mean((y_true - y_pred) / y_true) * 100
    nrmse = rmse / (y_true.max() - y_true.min())

    return {
        "RMSE": rmse,
        "MAE": mae,
        "NRMSE": nrmse,
        "MAPE": mape,
        "MPE": mpe,
    }


def train_vaticinator_variant(
    X_train,
    y_train,
    X_test,
    lags,
    epochs=100,
    batch_size=16,
    lr=1e-4,
    seed=42
):
    set_seed(seed)

    X_train_t = torch.tensor(
        X_train, dtype=torch.float32, device=DEVICE
    ).unsqueeze(-1)

    y_train_t = torch.tensor(
        y_train, dtype=torch.float32, device=DEVICE
    )

    X_test_t = torch.tensor(
        X_test, dtype=torch.float32, device=DEVICE
    ).unsqueeze(-1)

    model = Vaticinator(
        lags=lags,
        d_model=CONFIG["d_model"],
        nhead=CONFIG["nhead"],
        num_layers=CONFIG["num_layers"],
        dim_feedforward=CONFIG["dim_feedforward"]
    ).to(DEVICE)

    loader = DataLoader(
        TensorDataset(X_train_t, y_train_t),
        batch_size=batch_size,
        shuffle=False
    )

    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    losses = []

    for _ in range(epochs):
        model.train()
        epoch_loss = 0.0

        for xb, yb in loader:
            optimizer.zero_grad()
            pred = model(xb)
            loss = criterion(pred, yb)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item() * len(xb)

        losses.append(epoch_loss / len(loader.dataset))

    model.eval()
    with torch.no_grad():
        pred_scaled = model(X_test_t).cpu().numpy()

    pred_actual = scaler.inverse_transform(
        pred_scaled.reshape(-1, 1)
    ).flatten()

    actual = test["Total Revenue"].values

    return {
        "model": model,
        "losses": losses,
        "predictions": pred_actual,
        "metrics": regression_metrics(actual, pred_actual)
    }


primary_result = train_vaticinator_variant(
    X_train_combined,
    y_train_combined,
    X_test_combined,
    combined_lags,
    epochs=CONFIG["epochs"],
    batch_size=CONFIG["batch_size"],
    lr=CONFIG["lr"],
    seed=SEED
)

primary_result["metrics"]
```

## 8. Recursive Forecasting

```python
def recursive_forecast(model, train_scaled, lags, forecast_horizon):
    model.eval()
    history = list(np.asarray(train_scaled, dtype=np.float32))
    preds = []

    for _ in range(forecast_horizon):
        x = np.array(
            [[history[-lag] for lag in lags]],
            dtype=np.float32
        )

        x_t = torch.tensor(
            x, dtype=torch.float32, device=DEVICE
        ).unsqueeze(-1)

        with torch.no_grad():
            yhat = float(model(x_t).cpu().item())

        preds.append(yhat)
        history.append(yhat)

    return np.asarray(preds)


recursive_scaled = recursive_forecast(
    primary_result["model"],
    train_scaled,
    combined_lags,
    len(test)
)

recursive_pred = scaler.inverse_transform(
    recursive_scaled.reshape(-1, 1)
).flatten()

recursive_metrics = regression_metrics(
    actual_test,
    recursive_pred
)

recursive_metrics
```

## 9. Dense Transformer Baseline

```python
class StandardSequentialPE(nn.Module):
    def __init__(self, d_model, max_len=512):
        super().__init__()

        position = torch.arange(
            max_len, dtype=torch.float32
        ).unsqueeze(1)

        div_term = torch.exp(
            torch.arange(0, d_model, 2).float()
            * (-math.log(10000.0) / d_model)
        )

        pe = torch.zeros(max_len, d_model)
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)

        self.register_buffer("pe", pe.unsqueeze(0))

    def forward(self, x):
        return x + self.pe[:, :x.size(1)]


class DenseTransformer(nn.Module):
    def __init__(
        self,
        window_size=24,
        d_model=32,
        nhead=8,
        num_layers=3,
        dim_feedforward=128
    ):
        super().__init__()

        self.embedding = nn.Linear(1, d_model)
        self.pe = StandardSequentialPE(d_model, max_len=window_size)

        layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            activation="relu",
            batch_first=True
        )

        self.encoder = nn.TransformerEncoder(
            layer, num_layers=num_layers
        )

        self.output = nn.Linear(d_model, 1)

    def forward(self, x):
        x = self.embedding(x)
        x = self.pe(x)
        x = self.encoder(x)
        return self.output(x[:, -1, :]).squeeze(-1)
```

```python
X_train_dense, y_train_dense = create_aligned_lag_samples(
    train_scaled,
    dense_lags,
    start_index=24
)

X_test_dense = create_test_lag_samples(
    train_scaled,
    test_scaled,
    dense_lags
)

print("Dense train shape:", X_train_dense.shape)
print("Dense test shape :", X_test_dense.shape)
```

```python
def train_dense_transformer(
    X_train,
    y_train,
    X_test,
    epochs=100,
    batch_size=16,
    lr=1e-4,
    seed=42
):
    set_seed(seed)

    X_train_t = torch.tensor(
        X_train, dtype=torch.float32, device=DEVICE
    ).unsqueeze(-1)

    y_train_t = torch.tensor(
        y_train, dtype=torch.float32, device=DEVICE
    )

    X_test_t = torch.tensor(
        X_test, dtype=torch.float32, device=DEVICE
    ).unsqueeze(-1)

    model = DenseTransformer(
        window_size=24,
        d_model=CONFIG["d_model"],
        nhead=CONFIG["nhead"],
        num_layers=CONFIG["num_layers"],
        dim_feedforward=CONFIG["dim_feedforward"]
    ).to(DEVICE)

    loader = DataLoader(
        TensorDataset(X_train_t, y_train_t),
        batch_size=batch_size,
        shuffle=False
    )

    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    losses = []

    for _ in range(epochs):
        model.train()
        total = 0.0

        for xb, yb in loader:
            optimizer.zero_grad()
            pred = model(xb)
            loss = criterion(pred, yb)
            loss.backward()
            optimizer.step()
            total += loss.item() * len(xb)

        losses.append(total / len(loader.dataset))

    model.eval()
    with torch.no_grad():
        pred_scaled = model(X_test_t).cpu().numpy()

    pred_actual = scaler.inverse_transform(
        pred_scaled.reshape(-1, 1)
    ).flatten()

    return {
        "model": model,
        "losses": losses,
        "predictions": pred_actual,
        "metrics": regression_metrics(actual_test, pred_actual)
    }


dense_result = train_dense_transformer(
    X_train_dense,
    y_train_dense,
    X_test_dense,
    epochs=CONFIG["epochs"],
    batch_size=CONFIG["batch_size"],
    lr=CONFIG["lr"],
    seed=SEED
)

dense_result["metrics"]
```

### 10.4 LSTM

```python
class LSTMForecaster(nn.Module):
    def __init__(
        self,
        input_size=1,
        hidden_size=64,
        num_layers=2,
        dropout=0.2
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True
        )

        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        output, _ = self.lstm(x)
        return self.fc(output[:, -1, :]).squeeze(-1)
```

```python
X_train_lstm = X_train_dense[:, ::-1].copy()
X_test_lstm = X_test_dense[:, ::-1].copy()

X_train_lstm_t = torch.tensor(
    X_train_lstm, dtype=torch.float32, device=DEVICE
).unsqueeze(-1)

y_train_lstm_t = torch.tensor(
    y_train_dense, dtype=torch.float32, device=DEVICE
)

X_test_lstm_t = torch.tensor(
    X_test_lstm, dtype=torch.float32, device=DEVICE
).unsqueeze(-1)

set_seed(SEED)

lstm_model = LSTMForecaster().to(DEVICE)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(lstm_model.parameters(), lr=1e-3)

loader = DataLoader(
    TensorDataset(X_train_lstm_t, y_train_lstm_t),
    batch_size=16,
    shuffle=False
)

lstm_losses = []

for _ in range(300):
    lstm_model.train()
    total = 0.0

    for xb, yb in loader:
        optimizer.zero_grad()
        pred = lstm_model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        total += loss.item() * len(xb)

    lstm_losses.append(total / len(loader.dataset))

lstm_model.eval()

with torch.no_grad():
    lstm_pred_scaled = lstm_model(
        X_test_lstm_t
    ).cpu().numpy()

lstm_pred = scaler.inverse_transform(
    lstm_pred_scaled.reshape(-1, 1)
).flatten()

lstm_metrics = regression_metrics(actual_test, lstm_pred)
lstm_metrics
```

## 12. Positional Encoding Comparison

```python
class StandardPEVaticinator(Vaticinator):
    def __init__(
        self,
        lags,
        d_model=32,
        nhead=8,
        num_layers=3,
        dim_feedforward=128
    ):
        super().__init__(
            lags=lags,
            d_model=d_model,
            nhead=nhead,
            num_layers=num_layers,
            dim_feedforward=dim_feedforward
        )

        self.standard_pe = StandardSequentialPE(
            d_model=d_model,
            max_len=len(lags)
        )

    def forward(self, x):
        x = self.embedding(x)
        x = self.standard_pe(x)
        x = self.encoder(x)
        return self.output(x[:, -1, :]).squeeze(-1)
```

## 13. Sparse vs Dense Efficiency

```python
def measure_inference_time(model, X, warmup=50, repeats=500):
    model.eval()

    with torch.no_grad():
        for _ in range(warmup):
            _ = model(X)

    if torch.cuda.is_available():
        torch.cuda.synchronize()

    times = []

    with torch.no_grad():
        for _ in range(repeats):
            start = time.perf_counter()
            _ = model(X)

            if torch.cuda.is_available():
                torch.cuda.synchronize()

            times.append(
                (time.perf_counter() - start) * 1000
            )

    return {
        "mean_ms": float(np.mean(times)),
        "median_ms": float(np.median(times))
    }
```

```python
sparse_input = torch.tensor(
    X_test_combined,
    dtype=torch.float32,
    device=DEVICE
).unsqueeze(-1)

dense_input = torch.tensor(
    X_test_dense,
    dtype=torch.float32,
    device=DEVICE
).unsqueeze(-1)

sparse_timing_live = measure_inference_time(
    primary_result["model"],
    sparse_input
)

dense_timing_live = measure_inference_time(
    dense_result["model"],
    dense_input
)

print("Live sparse timing:", sparse_timing_live)
print("Live dense timing :", dense_timing_live)
```

# Final ESA Takeaway

## What the base paper established
Vaticinator showed that statistically guided lag-restricted Transformer input can remain competitive while reducing computational overhead and directional bias.

## What our replication adds
Our larger-data replication confirms the efficiency value of sparse temporal modelling, but exposes the key limitation:

> **The effectiveness of sparsity depends critically on which historical lags are selected.**

### Direct evidence
- **9-lag sparse:** RMSE 281.06M
- **Dense 24-token Transformer:** RMSE 271.34M
- **Compact 4-lag sparse:** RMSE **266.93M**

## Strong Phase-1 conclusion

> **Sparse temporal modelling is not automatically better simply because it is sparse. A compact, well-selected historical representation can outperform dense input, whereas a larger sparse set can perform worse. Therefore, lag relevance should not be assumed globally fixed.**

## Why Phase 2 follows logically

Phase 2 investigates:

> **Context-Conditioned Adaptive Temporal Relevance**

Instead of using one lag set for every forecast, the model will learn which historical periods are most relevant to the current product/store/context while preserving sparsity, efficiency and interpretability.
